```python
##Matplotlib
###Bar chart of membership types

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
##Bar chart of membership types

gym = gym['abonoment_type'].value_counts()

fig, ax = plt.subplots()
ax.bar(gym.index, gym.values, color = ['darkred', 'darkgreen'])
ax.set(title = 'Membership Types', ylabel = 'Count')
plt.show()
```

```python
###Histogram of avg_time_in_gym

DATA_DIR = '/content/drive/MyDrive/Datasets/'
gym = pd.read_csv(f'{DATA_DIR}/gym_membership.csv')

fig, ax = plt.subplots()
ax.hist(gym['avg_time_in_gym'], bins = 20, color = 'darkblue', edgecolor = 'white')
ax.set(title = 'Average Time in Gym', xlabel = 'Average Time', ylabel = 'Gym Time')
plt.show()
```

```python
##Scatter plot of Age vs visit_per_week

fig, ax = plt.subplots()
ax.scatter(gym['Age'], gym['visit_per_week'], alpha = 0.4, s =18)
ax.set(title = 'Visit Per Week vs Age', xlabel = 'Age', ylabel = 'Visit Per Week' )
plt.show()
```

```python
##Seaborn
###Box plot — gender vs Age

fig, ax = plt.subplots()
sns.boxplot(x = 'gender', y = 'Age', data = gym, ax = ax)
ax.set(title='Gender vs Age', xlabel = 'Gender', ylabel = 'Age')
plt.show()
```

```python


fig, axes = plt.subplots(1, 2, figsize = (11, 4))
sns.boxplot(x = 'gender', y = 'Age', data = gym, ax = axes[0])
axes[0].set(title='Gender vs Age', xlabel = 'Gender', ylabel = 'Age')

sns.boxplot(x = 'gender', y = 'visit_per_week', data = gym, ax = axes[1])
axes[1].set(title='Gender vs Visit Per Week', xlabel = 'Gender', ylabel = 'Visit Per Week')
plt.show()
```

```python
###Count plot — days_per_week

fig, ax = plt.subplots()
sns.countplot(x = 'days_per_week', data = gym)
ax.set(title='Days Per Week', xlabel = 'Days', ylabel = 'Count')

plt.tight_layout()
plt.show()
```

```python
###Bar plot — mean visit_per_week per abonoment_type

fig, ax = plt.subplots(figsize = (9, 4))
order = gym.groupby('abonoment_type')['visit_per_week'].mean().sort_values().index
sns.barplot(x = 'abonoment_type', y = 'visit_per_week', data = gym, order = order, ax = ax)

ax.set(title='Mean Visit Per Week per Abonoment Type')
ax.tick_params(axis = 'x', rotation = 30)
plt.tight_layout()
plt.show()
```
