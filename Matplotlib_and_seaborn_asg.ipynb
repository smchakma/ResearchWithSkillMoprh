```python
##Bar chart of membership types

gym = gym['abonoment_type'].value_counts()

fig, ax = plt.subplots()
ax.bar(gym.index, gym.values, color = ['darkred', 'darkgreen'])
ax.set(title = 'Membership Types', ylabel = 'Count')
plt.show()
```

```python
###Histogram of avg_time_in_gym

DATA_DIR = '/content/drive/MyDrive/Datasets/'
gym = pd.read_csv(f'{DATA_DIR}/gym_membership.csv')

fig, ax = plt.subplots()
ax.hist(gym['avg_time_in_gym'], bins = 20, color = 'darkblue', edgecolor = 'white')
ax.set(title = 'Average Time in Gym', xlabel = 'Average Time', ylabel = 'Gym Time')
plt.show()
```

```python
##Scatter plot of Age vs visit_per_week

fig, ax = plt.subplots()
ax.scatter(gym['Age'], gym['visit_per_week'], alpha = 0.4, s =18)
ax.set(title = 'Visit Per Week vs Age', xlabel = 'Age', ylabel = 'Visit Per Week' )
plt.show()
```
